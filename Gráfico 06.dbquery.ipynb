```sql
 SELECT
    `neighbourhood_cleansed` AS bairro,
    COUNT(*) AS quantidade_anuncios,
    ROUND(AVG(price), 2) AS preco_medio
FROM workspace.default.fact_listing_dimensional
WHERE price > 0
  AND review_scores_rating IS NOT NULL
GROUP BY `neighbourhood_cleansed`
ORDER BY quantidade_anuncios DESC
LIMIT 15
```
