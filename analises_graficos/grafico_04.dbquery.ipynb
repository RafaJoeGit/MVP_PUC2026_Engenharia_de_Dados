```sql
 SELECT
    room_type,
    COUNT(*) AS quantidade_anuncios,
    ROUND(AVG(price), 2) AS preco_medio,
    ROUND(AVG(availability_365), 1) AS disponibilidade_media_dias,
    ROUND(AVG(review_scores_rating), 2) AS avaliacao_media
FROM workspace.default.fact_listing_dimensional
WHERE price > 0
  AND review_scores_rating IS NOT NULL
GROUP BY room_type
ORDER BY avaliacao_media DESC;
```
